```python
import trimesh
```

```python
# 加载obj文件
mesh=trimesh.load("./obj/0_smpl.obj")

# 获取模型的点和面
vertices=mesh.vertices
faces=mesh.faces
```

vertices可以理解为二维数组，每一行代表一个点的坐标(x,y,z)
face是二维数组，每一行代表一个面，每个元素是一个点的索引

提取所有点的y坐标

```python
y_coords=vertices[:,1]
```

定义截取平面，像下面这样定义的y>0表示截取的就是y轴的上半部分

```python
y_threshold=(y_coords.min()+y_coords.max())/2
```

找到位于平面之上的点

```python
upper_vertices_indices=[i for i,v in enumerate(vertices) if v[1]>y_threshold]
```

找到包含这些顶点的面

```python
upper_faces=[]

for face in faces:
    # for vertex_index in face 找到一个面所有点的索引
    # 判断该点是否在upper_vertices_indices列表中，如果都在的话就把该面加入列表中
    if all(vertex_index in upper_vertices_indices for vertex_index in face):
        upper_faces.append(face)
```

创建新的顶点列表和面列表

```python
upper_vertices=vertices[upper_vertices_indices]
upper_faces=[[upper_vertices_indices.index(i) for i in face]for face in upper_faces]
```

创建新的mesh

```python
upper_mesh=trimesh.Trimesh(vertices=upper_vertices,faces=upper_faces)
```

保存截取后的模型

```python
upper_mesh.export("./test.obj")
upper_mesh.show()
```
