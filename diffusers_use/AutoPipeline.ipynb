下面，使用 dreamlike-art/dreamlike-photoreal-2.0 checkpoints

AutoPipeline做了如下的工作：

1.从model_index.json文件中检测"stable-diffusion" class

2.根据感兴趣的任务加载StableDiffusionPipeline, StableDiffusionImg2ImgPipeline, or StableDiffusionInpaintPipeline。任何想传递给这些管道的参数也可以传递给AutoPipeline

```python
import torch
from diffusers import AutoPipelineForText2Image

pipe_txt2img=AutoPipelineForText2Image.from_pretrained("dreamlike-art/dreamlike-photoreal-2.0", cache_dir="./cache",torch_dtype=torch.float16, use_safetensors=True).to("cuda:0")

prompt = "cinematic photo of Godzilla eating sushi with a cat in a izakaya, 35mm photograph, film, professional, 4k, highly detailed"

generator = torch.Generator(device="cuda:0").manual_seed(0)

image = pipe_txt2img(prompt, generator=generator).images[0]

image
```

AutoPipeline 支持 Stable Diffusion, Stable Diffusion XL, ControlNet, Kandinsky 2.1, Kandinsky 2.2, and DeepFloyd IF 的checkpoints。如果load一个不支持的checkpoints，会得到报错。
