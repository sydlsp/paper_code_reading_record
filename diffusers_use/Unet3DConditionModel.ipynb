```python
from diffusers.models import UNet3DConditionModel
```

```python
# cross_attention_dim要求和条件的token embedding维度是一致的
unet=UNet3DConditionModel(sample_size=64,cross_attention_dim=256).to("cuda:0")
```

```python
import torch
batch_size,channels,n_frames,h,w,len_squence=1,4,4,64,64,10

sample=torch.randn(size=[batch_size,channels,n_frames,h,w],dtype=torch.float32,device="cuda:0")

timestep=torch.randint(0,1000,size=[batch_size],device="cuda:0")

encoder_hidden_states=torch.randn(size=[batch_size,len_squence,256],dtype=torch.float32,device="cuda:0")
```

```python
out_put=unet(sample,timestep,encoder_hidden_states)
```
